Скачиваем пользовательские отзывы при помощи библиотеки google-play-scraper
Для данного проекта ставим лимит в 50 000 (произвольное число), язык английский, приложение ChatGPT, 
выбрано так как есть много интересных отзывов.

```python
# устанавливаем библиотеку для скачивания отзывов
!pip install google-play-scraper
```

```python
from google_play_scraper import reviews, Sort
import pandas as pd

all_reviews = []
continuation_token = None

for _ in range(250):  # 250 * 200 ≈ 50K
    result, continuation_token = reviews(
        'com.openai.chatgpt',
        lang='en',
        country='us',
        sort=Sort.NEWEST,
        count=200,
        continuation_token=continuation_token
    )

    all_reviews.extend(result)

    if not continuation_token:
        break

df = pd.DataFrame(all_reviews)
```

```python
# сохраняем корпус
df.to_csv('chatgpt_reviews_raw.csv', index=False)
```
