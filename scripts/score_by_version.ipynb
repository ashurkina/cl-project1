```python
import pandas as pd

df = pd.read_csv('../data/chatgpt_reviews_raw.csv')
df.head()
```

```python
df['date'] = pd.to_datetime(df['at']).dt.date

version_table = (
    df.dropna(subset=['appVersion'])
    .groupby('date')
    .agg(
        appVersion=('appVersion', lambda x: sorted(x.unique().tolist())),
        num_reviews=('score', 'count'),
        avg_score=('score', 'mean')
    )
    .reset_index()
    .sort_values('date')
)

version_table['avg_score'] = version_table['avg_score'].round(2)
version_table
```
